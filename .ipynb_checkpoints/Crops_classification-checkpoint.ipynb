from seeing the correlation we will found that the highest correlation with temperature column is humidity As N,K,P and pH are in the soil then we can apply dimensionality reduction PCA on it

```python
scaler = StandardScaler()
col_scale = ['ph','N','K','P']
X_scaled = scaler.fit_transform(df[col_scale])
pca = PCA(n_components=1)
df['soil fertility'] = pca.fit_transform(X_scaled)
df.head(10)
```

```python
df.columns
```

apply encoding

```python
label_encoder = LabelEncoder()
df['crops'] = label_encoder.fit_transform(df['crops'])
```

applying classification

```python
X2 = df.drop(['crops','N', 'P', 'K','ph'], axis=1)
y2 = df['crops']
X_train2, X_test2, y_train2, y_test2 = train_test_split(X2, y2, test_size=0.2, random_state=42)
```

apply scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler2 = StandardScaler()
X_train_scaled2 = scaler2.fit_transform(X_train2)
X_test_scaled2 = scaler2.transform(X_test2)
```

```python
df.head()
```

```python
param_grid2 = {'n_neighbors': range(1, 20)}
knn2 = KNeighborsClassifier()
grid_search2 = GridSearchCV(knn2, param_grid2) 
grid_search2.fit(X_train_scaled2, y_train2)
best_k2 = grid_search2.best_params_['n_neighbors']
print(f"Best value for k: {best_k2}")
```

```python
models = {
    'LogisticRegression': LogisticRegression(),
    'DecisionTreeClassifier': DecisionTreeClassifier(),
    'SVC': SVC(),
    'GaussianNB': GaussianNB(),
    'KNeighborsClassifier':KNeighborsClassifier(n_neighbors=best_k2),
    'BaggingClassifier': BaggingClassifier(),
    'AdaBoostClassifier':AdaBoostClassifier(),
    'GradientBoostingClassifier': GradientBoostingClassifier(),
    'XGBClassifier':XGBClassifier()
}
```

```python
models_acc = dict()
for m_name,m_model in models.items():
    m_model.fit(X_train_scaled2,y_train2)
    y_train_pred = m_model.predict(X_train_scaled2)
    train_acc = accuracy_score(y_train2, y_train_pred)
    y_test_pred = m_model.predict(X_test_scaled2)
    test_acc = accuracy_score(y_test2, y_test_pred)
    models_acc[m_name] = test_acc
    print(f"Train accuracy of {m_name}: {train_acc:.2f}")
    print(f"Test accuracy of {m_name}: {test_acc:.2f}")
```
